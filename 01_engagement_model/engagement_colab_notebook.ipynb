```python
# Cell 5: Test Real-time Engagement Inference
import joblib
import pandas as pd

model = joblib.load('models/engagement_xgb.joblib')
features = joblib.load('models/engagement_features.joblib')

sample_student = pd.DataFrame([{
    'total_clicks': 55,
    'active_days': 4,
    'unique_materials': 6,
    'clicks_2w_mean': 48.0,
    'active_days_2w': 7,
    'click_change': 12
}])[features]

prob = float(model.predict_proba(sample_student)[0, 1])
print(f'Predicted Engagement Probability for Next Week: {prob:.1%}')
```
